```python
import json
import os
from collections import defaultdict
from pathlib import Path

def analyze_image_quality_json_files(json_dir):
    """
    Analyze all JSON files in the directory and count occurrences of each attribute value.
    """
    
    # Initialize counters for each attribute
    counters = {
        'rotation': defaultdict(int),
        'clarity': defaultdict(int),
        'handwriting': defaultdict(int),
        'writing_style': defaultdict(int),
        'lighting': defaultdict(int),
        'resolution': defaultdict(int),
        'paper_condition': defaultdict(int),
        'text_alignment': defaultdict(int),
        'Strikethrough': defaultdict(int)
    }
    
    # Get all JSON files in the directory
    json_files = [f for f in os.listdir(json_dir) if f.endswith('.json')]
    total_files = len(json_files)
    
    print(f"Found {total_files} JSON files to analyze...")
    print("=" * 60)
    
    # Process each JSON file
    for filename in json_files:
        filepath = os.path.join(json_dir, filename)
        
        try:
            with open(filepath, 'r', encoding='utf-8') as file:
                data = json.load(file)
                
                # Extract values from the result section
                result = data.get('result', {})
                
                # Count each attribute value
                for attribute in counters.keys():
                    value = result.get(attribute, 'NOT_FOUND')
                    counters[attribute][value] += 1
                    
        except Exception as e:
            print(f"Error processing {filename}: {e}")
    
    # Display results
    print(f"\nANALYSIS RESULTS for {total_files} JSON files:")
    print("=" * 60)
    
    for attribute, counts in counters.items():
        print(f"\n{attribute.upper()}:")
        print("-" * 30)
        total_count = sum(counts.values())
        
        for value, count in sorted(counts.items()):
            percentage = (count / total_count) * 100 if total_count > 0 else 0
            print(f"  {value:25} : {count:4d} ({percentage:5.1f}%)")
        
        print(f"  {'TOTAL':25} : {total_count:4d}")
    
    # Summary statistics
    print("\n" + "=" * 60)
    print("SUMMARY STATISTICS:")
    print("=" * 60)
    
    # Rotation analysis
    rotation_counts = counters['rotation']
    print(f"ROTATION:")
    print(f"  Upright: {rotation_counts['upright']}")
    print(f"  Clockwise: {rotation_counts['clockwise']}")
    print(f"  Anti-clockwise: {rotation_counts['anti_clockwise']}")
    print(f"  Inverted: {rotation_counts['inverted']}")
    
    # Clarity analysis
    clarity_counts = counters['clarity']
    print(f"\nCLARITY:")
    print(f"  Clear: {clarity_counts['clear']}")
    print(f"  Slightly blur: {clarity_counts['slightly_blur']}")
    print(f"  Very blur: {clarity_counts['very_blur']}")
    print(f"  Incomprehensible: {clarity_counts['incomprehensible']}")
    
    # Handwriting analysis
    handwriting_counts = counters['handwriting']
    print(f"\nHANDWRITING:")
    print(f"  Good: {handwriting_counts['good']}")
    print(f"  Okay: {handwriting_counts['okay']}")
    print(f"  Bad: {handwriting_counts['bad']}")
    print(f"  Incomprehensible: {handwriting_counts['incomprehensible']}")
    
    # Writing style analysis
    writing_style_counts = counters['writing_style']
    print(f"\nWRITING STYLE:")
    print(f"  Neat and tidy: {writing_style_counts['neat_and_tidy']}")
    print(f"  Okay: {writing_style_counts['okay']}")
    print(f"  Messy: {writing_style_counts['messy']}")
    
    # Lighting analysis
    lighting_counts = counters['lighting']
    print(f"\nLIGHTING:")
    print(f"  Consistent: {lighting_counts['consistent']}")
    print(f"  Strong contrast shadows: {lighting_counts['strong_contrast_shadows']}")
    print(f"  Glare on text: {lighting_counts['glare_on_text']}")
    
    # Resolution analysis
    resolution_counts = counters['resolution']
    print(f"\nRESOLUTION:")
    print(f"  High: {resolution_counts['high']}")
    print(f"  Medium: {resolution_counts['medium']}")
    print(f"  Low: {resolution_counts['low']}")
    print(f"  Needs enhancement: {resolution_counts['needs_enhancement']}")
    
    # Paper condition analysis
    paper_condition_counts = counters['paper_condition']
    print(f"\nPAPER CONDITION:")
    print(f"  Clean: {paper_condition_counts['clean']}")
    print(f"  Creased: {paper_condition_counts['creased']}")
    print(f"  Torn: {paper_condition_counts['torn']}")
    print(f"  Folded: {paper_condition_counts['folded']}")
    
    # Text alignment analysis
    text_alignment_counts = counters['text_alignment']
    print(f"\nTEXT ALIGNMENT:")
    print(f"  Well aligned: {text_alignment_counts['well_aligned']}")
    print(f"  Slightly misaligned: {text_alignment_counts['slightly_misaligned']}")
    print(f"  Heavily misaligned: {text_alignment_counts['heavily_misaligned']}")
    print(f"  Overcrowded: {text_alignment_counts['overcrowded']}")
    
    # Strikethrough analysis
    strikethrough_counts = counters['Strikethrough']
    print(f"\nSTRIKETHROUGH:")
    print(f"  Present: {strikethrough_counts['present']}")
    print(f"  Absent: {strikethrough_counts['absent']}")
    
    return counters

if __name__ == "__main__":
    # Path to the JSON files directory
    json_directory = "/Users/simrannaik/Desktop/solution_improvement/ds-prototypes/subjective_grading/solution_improvement/z2_ocr_gemini_pcmb/trial/image_quality/Maths/json"
    
    # Run the analysis
    results = analyze_image_quality_json_files(json_directory)
```
